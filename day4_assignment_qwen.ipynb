```python
from groq import Groq
from google.colab import userdata
key = userdata.get('GROQ_API_KEY')

client = Groq(api_key=key)

print("Groq API connected successfully")
```

```python
models = client.models.list()
print("AVAILABLE GROQ MODELS:\n")
for model in models.data:
  print(model.id)
```

```python
response = client.chat.completions.create(
    model= "qwen/qwen3.8-27b",
    messages= [
        {"role": "system", "content": "You are an fast learner."},
        {"role": "user", "content": "explain CNN in 5 points and give 5 interview points based on CNN for 3rd year student."}
    ],
    temperature=0.7,
    max_tokens=1000
)
print(response.choices[0])
print(response.choices[0].message.content)
```

```python
response = client.chat.completions.create(
    model= "openai/gpt-oss-safeguard-20b",
    messages= [
        {"role": "system", "content": "You are an very good learner."},
        {"role": "user", "content": "give 5 soft skill interview questions for practice and also give 5 aptitude questions based on probability."}
    ],
    temperature=0.7,
    max_tokens=1000
)
print(response.choices[0])
print(response.choices[0].message.content)
```
