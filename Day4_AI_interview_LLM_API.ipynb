```python
from groq import Groq
from google.colab import userdata
key = userdata.get('GROQ_API_KEY')

client = Groq(api_key=key)

print("Groq API connected successfully")
```

```python
models = client.models.list()
print("AVAILABLE GROQ MODELS:\n")
for model in models.data:
  print(model.id)
```

Test a Prompt

```python
response = client.chat.completions.create(
    model= "openai/gpt-oss-120b",
    messages= [
        {"role": "system", "content": "You are an exerienced technical interview."},
        {"role": "user", "content": "generate 5 internediate-level python interview questions for final-year engineering students."}
    ],
    temperature=0.7,
    max_tokens=1000
)
print(response.choices[0])
print(response.choices[0].message.content)
```
