```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, MinMaxScaler
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
```

2.  Load datasets

```python
df = pd.read_csv('/content/CAR DETAILS FROM CAR DEKHO.csv')
df.head()
```

3.  Find null values and drop

```python
df.isnull().sum()
```

4. Find duplicates and drop

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace= True)
df.duplicated().sum()
```

5.  Apply label encoding on fuel and transmission columns

```python
FE = LabelEncoder()
df["fuel_encoder"] = FE.fit_transform(df["fuel"])
df.head(11)
```

```python
TE = LabelEncoder()
df["transmission_encoder"] = TE.fit_transform(df["transmission"])
df.head(11)
```

6. separate X and Y

```python
X = df[["year", "km_driven", "fuel_encoder","transmission_encoder"]]
Y = df["selling_price"]
print("feature shape:", X.shape, "Target shape:",Y.shape)
```

7. split train test (20%)

```python
X_train,X_test,Y_train,Y_test = train_test_split(
    X,Y, test_size = 0.20, random_state = 42)
print("Records for training:", len(X_train))
print("Records for testing:", len(X_test))
```

8. Apply minmax scaler on Xtest and Xtrain


```python
num_cols = ["year", "km_driven", "fuel_encoder","transmission_encoder"]
my_scaler = MinMaxScaler()
X_train[num_cols] = my_scaler.fit_transform(X_train[num_cols])
X_test[num_cols] = my_scaler.transform(X_test[num_cols])
print(X_train)
```

9. Create linear regression model and fit with Xtrain Aand Ytrain

```python
selling_model = LinearRegression()
selling_model.fit(X_train,Y_train)
```

10. Predict selling prices for X test and create a comparison dataframe with Actual and Predicted prices

```python
y_pred = selling_model.predict(X_test)
result = pd.DataFrame(
    {
        "Actual" : Y_test,
        "Prediction_by_model" : y_pred
    }
)
result.head(11)
```
