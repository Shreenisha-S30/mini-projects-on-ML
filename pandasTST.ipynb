```python
df = pd.read_csv("/content/Student_Performance_100_Notclean.csv")
print(df)

df.head(7)
```

```python
df.info()
```

```python
subjects_col = ["Python","Mathematics","DBMS", "AI", "DataStructures"]
pass_marks = 40
```

data cleaning

- null
-

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)  #inplace --> save changes in save data file
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace = True)
df.duplicated().sum()
```

total and avg marks - create column


```python
#df["new_col"] = expression
df["total"] = (df[subjects_col]).sum(axis = 1)
df["avg"] = df[subjects_col].mean(axis=1)
df.head()
```

```python
print("avg of avg:", df["avg"].mean().round(2))
```

```python
topper_index = df["total"].idxmax()
df.loc[topper_index]
print("Topper name:",df.loc[topper_index, "StudentName"])
```

```python
topper_index = df["total"].idxmax()
df.loc[topper_index]
```

find the top scorer in mathematics and print their name,ID and marks

```python
math_topper = df["Mathematics"].idxmax()
df.loc[math_topper]
print("Topper name:",df.loc[math_topper, "StudentName"])
print("topper id:",df.loc[math_topper, "StudentID"])
print("mathematics marks:",df.loc[math_topper,"Mathematics"])
```

print all the names whose total marks is more than 400


```python
total_marks = df["total"]>400
df.loc[total_marks]
print("names of students who scores above 400:\n ",df.loc[total_marks, "StudentName"])
```

find the least aaverage(name,id)

```python
lower_index = df["total"].idxmin()
df.loc[lower_index]
print("the least avg student name:",df.loc[lower_index,"StudentName"])
print("ID:",df.loc[lower_index,"StudentID"])
```

```python
math_topper = df["Mathematics"].idxmax()
df.loc[math_topper]
print("Topper in maths:\n",df.loc[math_topper,["StudentName", "StudentID", "Mathematics"]])
```

```python
total_marks = df[df["total"]>400]
print("names of students who scores above 400: ",total_marks['StudentName'])
```

create pass/fail column


```python
cond = df[subjects_col]>pass_marks
passed_rec = cond.all(axis = 1)
df["Result"] = passed_rec.map(
    {
        True : "Pass",
        False : "Fail"
})
df.head()
```
